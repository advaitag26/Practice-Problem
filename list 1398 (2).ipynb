```python
number = []
for i in range(10):
    num = int(input("Enter Value:"))
    number.append(num)

print("\nOriginal List :", number)

ascending = sorted(number)
print("Ascending Order: ", ascending)

number.sort(reverse = True)
print("Descending Order: ", number)

print("Length of List:", len(number))
```

```python
list1 = []
list2 = []

print("\nEnter the 5 elements of List 1:")

for i in range(5):
    num = int(input("Enter the values of list 1:"))
    list1.append(num)
print("\nEnter the 5 elements of List 2:")

for i in range(5):
    num = int(input("Enter the value of list 2:"))
    list2.append(num)

print("LIST 1:" , list1)
print("LIST 2:" , list2)

print("Merged List:", list1+ list2)
```

```python
phrase = input("Enter a phrase:")
words = phrase.split()
acronym = ""
for word in words:
    acronym = acronym + word[0]
print("Acronym :", acronym.upper())
```
