```python
#Data Frames
import pandas as pd
data = {
    "calories":[420,380,390],
    "duration":[50,40,45]
}
myvar = pd.DataFrame(data)
print(myvar)
```

```python
#Data cleaning and preprocessing
import pandas as pd
data ={
    "A":[1,2,3,None,5],
    "B":[None,2,3,4,5],
    "C":[1,2,None,None,5],
}
df=pd.DataFrame(data)
print("Original Data:\n",df)

df_cleaned = df.dropna()
print("\ncleaned Data:\n",df_cleaned)
```

```python
#Data cleaning and preprocessing2
import pandas as pd
data ={
    "A":[1,2,3,None,5],
    "B":[None,2,3,4,5],
    "C":[1,2,None,None,5],
}
df=pd.DataFrame(data)
print("Original Data:\n",df)

df.fillna(0,inplace=True)
print("\nUpdated Data:\n",df)
```

```python
#Data cleaning and preprocessing3
import pandas as pd
data ={
    "A":[1,2,3,None,5],
    "B":[None,2,3,4,5],
    "C":[1,2,None,None,5],
}
df=pd.DataFrame(data)
print("Original Data:\n",df)

df.fillna(df.mean(),inplace = True)
print("\nUpdated Data:\n",df)
```
