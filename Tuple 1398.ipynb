```python
num= (10,20,30,40,50)
element = int(input("Enter the element:"))
if element in num :
    print("Index of element:",num.index(element))
else:
    print("Element not found.")
```

```python
# list tuple eddition
num = [(10,20,30),(40,50,60),(70,80,90)]
new_list=[]
for tup in num:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)
print("Original List:", num)
print("Updated List:", new_list)
```

```python
#max min 
num = (25 ,10,3,34,45)
maximum = max(num)
minimum = min(num)
print("Tuple :",num)
print("Maximim element:",maximum)
print("Minimum element:",minimum)
```

```python
#tuple swaping
num = (10,20,30,40,50,60)
new_tup= (num[-1],) +num[1:-1] + (num[0],)
print("Original Tuple:",num)
print("Tuple after swapping:",new_tup)
```
